Como a quantidade de filmes presentes no IMDb variou ao longo dos anos?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv("imdb_top_1000.csv")

#df.info()
df["Released_Year"]=pd.to_numeric(df["Released_Year"], errors="coerce")

df = df.dropna(subset=["Released_Year"])
filmes_por_ano = df["Released_Year"].value_counts().sort_index()
#df.info()

plt.bar(filmes_por_ano.index,  filmes_por_ano.values, color="red")
plt.title("Quantidade de filmes lançados por ano(segundo IMDB)")
plt.xlabel("Ano de lançamento")
plt.ylabel("Quantidade de filmes")
plt.show()
```

# Conclusão!!
O grafico aponta a variação da quantidade de filmes ao longo dos anos, indicando períodos com maior empenho de lançamentos. Nota-se que,  há um aumento no número de filmes presentes no conjunto de dados, indicando uma maior produção ou maior reconhecimento desses períodos.

## Observação:
Os dados representam apenas os filmes que estão no conjunto analisado, e não a tonalidade dos filmes lançados mundialmente em cada ano.

# Pergunta nº2
Qual classificação indicativa aparece com mais frequência no IMDb?

```python

import pandas as pd
import matplotlib.pyplot as plt



df=pd.read_csv("imdb_top_1000.csv")
df = df.dropna(subset=["Certificate"])
certificados= df["Certificate"].value_counts()
plt.bar(certificados.index, certificados.values, color="salmon")
plt.title("Frequencia das classificações indicativas de acordo com o IMDB")
plt.xlabel("Classificação indicativa")
plt.ylabel("Frequência")
plt.show()
```

## Conclusão!!
O grafico aponta que algumas classificações indicativas aparecem mais vezes entre os filmes analisados. Por outro lado, isso não significa que a classificação seja responsável pela qualidade e popularização dos filmes.

## Observação:
Foi visto que algumas classificações aparecem mais vezes, o que pode ter uma relação ao tipo de filmes mais comuns nesse conjunto de dados.
